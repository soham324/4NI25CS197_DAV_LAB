### Visualization of Rating and Budget
We will now look at the distribution of student ratings and the budget allocations.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram of Final_Score (substituting for missing 'Rating')
plt.figure(figsize=(8, 5))
if 'Final_Score' in df.columns:
    sns.histplot(df['Final_Score'], bins=10, kde=True, color='skyblue')
    plt.title('Histogram of Final Scores')
    plt.xlabel('Final Score')
    plt.ylabel('Frequency')
else:
    print("Column 'Final_Score' not found.")
plt.show()
```

```python

plt.figure(figsize=(8, 5))
if 'Attendance' in df.columns:
    sns.boxplot(y=df['Attendance'], color='lightgreen')
    plt.title('Box Plot of Attendance')
    plt.ylabel('Attendance Percentage')
else:
    print("Column 'Attendance' not found.")
plt.show()
```
